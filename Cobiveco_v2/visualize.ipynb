```python
import numpy as np
import matplotlib.pyplot as plt

# --- 1. Load Data ---
file_path = './DIMON_training_data_healthy.npz'
data = np.load(file_path)

coords = data['reference_coords']  # Shape: (N, 3)
u_data = data['u_data']            # Shape: (num_cases, num_sims, N)

# Extract spatial coordinates
x = coords[:, 0]
y = coords[:, 1]
z = coords[:, 2]

# Extract simulation values for coloring (Case 0, Simulation 0)
colors = u_data[0, 2, :]

# --- 2. Plotting ---
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Scatter plot with size 's=1' to handle dense point clouds gracefully
scatter = ax.scatter(x, y, z, c=colors, cmap='jet', s=1, alpha=0.8)

# Formatting
cbar = fig.colorbar(scatter, ax=ax, shrink=0.5, pad=0.1)
cbar.set_label('Simulation Value')

ax.set_title('3D Point Cloud (Case 0, Sim 0)')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')

# Ensure correct anatomical proportions
ax.set_box_aspect((np.ptp(x), np.ptp(y), np.ptp(z)))

plt.show()
```
