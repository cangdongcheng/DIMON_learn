```python
import numpy as np

# Load the dataset
data_path = "./dimon_training_data.npz"
dataset = np.load(data_path)

print(f"--- Dataset Inspection: {data_path} ---")

# Iterate through all keys and print their shapes
for key in dataset.files:
    data = dataset[key]
    print(f"Key: {key:20} | Shape: {str(data.shape):15} | Type: {data.dtype}")

# Optional: Print a small sample of the coordinates or pacing to verify content
if 'pacing' in dataset:
    print(f"\nPacing Sample (First 2):\n{dataset['pacing'][:2]}")

print("---------------------------------------")
```
